```python
import numpy as np

sales = [
    [120,130,125,140],
    [135,128,138,145],
    [142,150,147,155],
    [150,160,158,162],
    [170,168,175,180],
    [185,190,188,195]
]

sales_array = np.array(sales)

print("Sales Array:")
print(sales_array)

monthly_sales = np.sum(sales_array, axis=0)
print("\nTotal Sales of Each Month:")
print(monthly_sales)

product_sales = np.sum(sales_array, axis=1)
print("\nTotal Sales of Each Product:")
print(product_sales)

highest_month = np.argmax(monthly_sales) + 1
print("\nMonth with Highest Sales:", highest_month)

updated_sales = sales_array * 1.10
print("\nUpdated Sales (10% Increase):")
print(updated_sales)
```

```python
import pandas as pd

data = {
    "ID":[101,102,103,104,105,106],
    "Name":["Amit","Priya","Rahul","Sneha","Kiran","Ravi"],
    "Department":["HR","IT","Finance","IT","HR","Finance"],
    "Salary":[35000,50000,45000,55000,40000,60000],
    "Experience":[2,5,4,6,3,8]
}

df = pd.DataFrame(data)

print("First 3 Rows:")
print(df.head(3))

print("\nIT Department Employees:")
print(df[df["Department"] == "IT"])

print("\nAverage Salary:")
print(df["Salary"].mean())


print("\nSorted by Salary:")
print(df.sort_values(by="Salary", ascending=False))

df["Bonus"] = df["Salary"] * 0.10

print("\nDataFrame with Bonus:")
print(df)
```
